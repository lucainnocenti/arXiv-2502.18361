## 2024-09-02 data

Input states:
- 150 separable states, spanning a 9-dimensional subspace
- 150 entangled, spanning a 6-dimensional subspace

Separable and entangled states together span a *10*-dimensional subspace. Meaning that the task of learning arbitrary properties of entangled states training only on separables is here theoretically impossible, as the training doesn't have sufficient information (because entangled states cover  directions in state space not covered by the separables). This is due to this specific way of generating states, it's not a fundamental limitation.

All measurement data related to each state is taken sequentially. Meaning that the *time-arrow* follows the *state number*, rather than the repetition number.

## 2024-09-30 data

Complete set of training/test states (using different angles to generates states).

*States used*: 300 entangled states, 150 separable states with HV input, 150 separable states with VH input.
Same (randomly sampled) angles used to generate both entangled and separable datasets.

```python
check_singular_values_states(dir_experimental_data_0930, which_labels=['sep', 'ent', 'all'])
```

```python
from QELM import kets_to_vectorized_states_matrix
def _():
    separable_states = kets_to_vectorized_states_matrix(ExperimentalDataset(dir_experimental_data_0930).get_training_dataset(
        which_states='sep', which_counts='doubles',
        target_observables=two_qubit_paulis, save_states=True
    )['states'])

    entangled_states = kets_to_vectorized_states_matrix(ExperimentalDataset(dir_experimental_data_0930).get_training_dataset(
        which_states='sep', which_counts='doubles',
        target_observables=two_qubit_paulis, save_states=True
    )['states'])

    all_states = kets_to_vectorized_states_matrix(ExperimentalDataset(dir_experimental_data_0930).get_training_dataset(
        which_states='all', which_counts='doubles',
        target_observables=two_qubit_paulis, save_states=True
    )['states'])

    print(f'Singular values for separable states:\n', np.linalg.svd(separable_states, compute_uv=False))
    print(f'Singular values for entangled states:\n', np.linalg.svd(entangled_states, compute_uv=False))
    print(f'Singular values for all states:\n', np.linalg.svd(all_states, compute_uv=False))
_()
```
